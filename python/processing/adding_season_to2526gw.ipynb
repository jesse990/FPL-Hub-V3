```python
import pandas as pd
from pathlib import Path

# ====================== CONFIG ======================
folder_path = Path(r"C:\Users\JesseOnu\fpl sql rework\gws")   # ← CHANGE THIS to your folder
season = 2526
# ===================================================

# Find all CSV files matching the pattern 2526gw*.csv
csv_files = list(folder_path.glob("2526gw*.csv"))

print(f"Found {len(csv_files)} CSV files.")

for file_path in sorted(csv_files):
    try:
        # Read the CSV
        df = pd.read_csv(file_path)
        
        # Add the season column
        df['season'] = season
        
        # Save back to the same file
        df.to_csv(file_path, index=False)
        
        print(f"✓ Updated: {file_path.name}  |  Rows: {len(df):,}")
        
    except Exception as e:
        print(f"✗ Error processing {file_path.name}: {e}")

print("\nAll done!")
```
